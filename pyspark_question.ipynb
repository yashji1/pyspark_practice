Q1 What is Apache Spark and how does PySpark relate to it?

> Apache Spark is a distributed data processing engine used for large-scale data processing (batch + streaming) across clusters.

> PySpark is not a separate framework, but a Python API/interface that allows you to write Spark programs using Python instead of Scala/Java.

> PySpark doesn’t “run on top” independently—it acts as a bridge to the Spark engine, which actually executes everything on the cluster.

Q2. What are the differences between:

RDD
DataFrame
Dataset

- RDD (Resilient Distributed Dataset)
    - Low-level abstraction
    - Works with unstructured data
    - No schema (no column names/types)
    - Slower because no optimization

- DataFrame
    - Higher-level abstraction (like a table)
    - Works with structured + semi-structured data
    - Has schema (columns + types)
    - Faster due to Catalyst Optimizer & Tungsten execution

- Dataset
    - Type-safe version of DataFrame
    - Mostly used in Scala/Java, not in PySpark
    - Combines benefits of RDD + DataFrame

Q3. What is lazy evaluation in Spark?

> Lazy evaluation in Apache Spark means that transformations are not executed immediately. Instead, Spark builds a logical execution plan (DAG), and only when an action like collect() or show() is called, the entire plan is executed.

Q4. What is the difference between:

Transformations   
Actions

- Transformations:    
    - Operations that define how data should be changed
    - They are lazy (not executed immediately)
    - Return a new RDD/DataFrame.

    Examples:

    - filter()
    - map()
    - select()
    - groupBy()


- Actions     
    - Operations that trigger execution
    - They return results to driver or write output

    Examples:

    - collect() → brings data to driver  
    - show() → displays data   
    - count() → returns number of rows  
 

Q5. What is the difference between:

map()
flatMap()

> map() returns one output per input (can be nested), while flatMap() flattens the output and can return multiple elements per input.

Q6. What happens when you run collect() on a large dataset?
(Think about memory and driver node )

> collect() brings all distributed data to the driver node. This is dangerous for large datasets because it can cause memory overflow on the driver and degrade performance.

(Hint: Catalyst Optimizer)

> Spark uses the Catalyst Optimizer to optimize queries. It converts code into a logical plan, applies optimization rules like predicate pushdown and column pruning, then generates an efficient physical plan for execution.

Q10. What types of joins are supported in PySpark?

> PySpark supports inner, left, right, full outer, left semi, left anti, and cross joins. Broadcast join is not a join type but an optimization technique used for small datasets.

Q11. What is a broadcast join and when would you use it?

> A broadcast join is an optimization where a small dataset is sent (broadcasted) to all worker nodes, so that the join can happen locally on each node without shuffling the large dataset.

Q12. What is data skew and how do you handle it?
