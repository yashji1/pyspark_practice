# **(RDD - Resilent Distributed Data)**
RDD is used when data is unstructed form.
- RDD is just a collection of object
- it does not contain any schema.
- it uses python functions.

`rdd = spark.sparkContext.parallelize(data)`

- rdd uses python function
`rdd.filter(lambda x: x[2] > 5000)`
- python code runs outside the JVM.
- spark cannot inspect our logic. Kills performace.



Note: Cannot use sparkContext with serverless so dataframes is better.

# DataFrame
Use when data is structured.
- understand the operations.
- ability to read schema.
- picks best plan execution.
`df = spark.createDataFrame(data, schema)`
- dataframes operations are declarative.`df.filter(df.salary > 5000)`


```python
schema = ['name', 'age', 'department', 'salary']
```

```python
df = spark.createDataFrame(data, schema)
display(df)
```

Note: Both have lazy evalution.
- Dataframe builds optimize logic plan.
- rdd just chain function calling.

# Repartition() and Coalese()

- repartition can increase or decrease the number of partition.
- coalese can only decrease the number of partition.

```python
student_df = spark.read.format('csv') \
    .option("inferedSchema", "true") \
    .option("header", "true") \
    .load("/Workspace/Users/sharmayashtdl@gmail.com/pyspark_practice/student_data_223_rows.csv")
```

- repartiton divides data into equal parts 
- coalese jsut combine the other partitions.

```python
student_df.show(5)
```

```python
partitioned_stud_df = student_df.repartition(4)
```

it divides the data in a equal partitions.

```python
partitioned_stud_df.withColumn("partition_id", spark_partition_id()).groupBy("partition_id").count().show()
```

```python
partition_one_column = student_df.repartition(300,'Marks')
```

if we have more number of partiton it skips some partiton for utilize.

```python
partition_one_column.withColumn("partition_id", spark_partition_id()).groupBy("partition_id").count().show(30)
```
