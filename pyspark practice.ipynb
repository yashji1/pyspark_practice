# **(RDD - Resilent Distributed Data)**
RDD is used when data is unstructed form.
- RDD is just a collection of object
- it does not contain any schema.
- it uses python functions.

`rdd = spark.sparkContext.parallelize(data)`

- rdd uses python function
`rdd.filter(lambda x: x[2] > 5000)`
- python code runs outside the JVM.
- spark cannot inspect our logic. Kills performace.



Note: Cannot use sparkContext with serverless so dataframes is better.

# DataFrame
Use when data is structured.
- understand the operations.
- ability to read schema.
- picks best plan execution.
`df = spark.createDataFrame(data, schema)`
- dataframes operations are declarative.`df.filter(df.salary > 5000)`


```python
schema = ['name', 'age', 'department', 'salary']
```

```python
df = spark.createDataFrame(data, schema)
display(df)
```

Note: Both have lazy evalution.
- Dataframe builds optimize logic plan.
- rdd just chain function calling.
